</div>

Understand what you want to accomplish. This is the first step of the CRISP-DM framework for data science projects. Business understanding is about looking at the data in context. Skipping this step could lead to projects that are intensive work but lack any real meaning or impact in the world the data came from or in the organization looking for insight from it.

**a. Read the dataset description. Who produced tha data? When, how, and for what purpose?**

**Answer:** The data is produced and maintained by New York City government agencies (via the NYC Open Data portal and managed through NYC service operations). It is collected from 2020 to the present day through public submissions via phone calls, mobile apps, and the web. The purpose of the data is to log, track, and manage public service requests and complaints from citizens regarding rodent sightings and conditions and allowing city agencies to address public health and sanitation issues.

**b. Name a plausible stakeholder, their goals, and the concrete decision the analysis would support. Who benefits, who could be harmed?**

**Answer:** 
Stakeholder & Goal: The New York City Department of Health and Mental Hygiene or local pest control management teams, whose goal is to efficiently allocate inspection resources and extermination treatments.

Concrete Decision: Deciding where to deploy inspection teams and target high risk neighborhoods based on predictive hotspot concentrations.

Who benefits: City residents and business owners (like restaurants) benefit from reduced rat populations, improved health standards, and quicker response times to infestations.

Who could be harmed: Property owners or landlords in areas incorrectly flagged as high risk might face fines, inspections, or negatively affect property values.

**c. State the objective in business terms (decide strategy, predict demand or delay, recognize priorities, connect factors to outcomes, detect risks, etc), and how success would be judged (cost saved, error tolerated, etc).**

For example:
* _"Predict delay for bus lines so that passengers can choose an earlier connection."_
* _"Recognize priorities among customer complaints so that support staff answer urgent ones first."_

NB! For this step, remember to name the objective without naming the method. For example, don't state “predict X with a random forest” as the objective.

**Answer:** 
Business Objective: Recognize priorities and predict future concentrations of rat complaints so that city sanitation and pest control teams can deploy preventative interventions before infestations peak.

Success Criteria: Success would be judged by a reduction in repeat complaints in targeted zones, optimized deployment times, and minimizing the cost of wasted inspections in low-activity areas while maintaining a low false-negative rate for critical hot spots.

**d. Now translate the objective to a data science problem.**

Is it classification, regression, clustering, description? What would be predicted or explained? At what point in time would the prediction or other outcome be used?

**Answer:** 
It is both description (analyzing historical patterns of where and when rats are reported) and prediction/regression (predicting future concentrations or probability of high-complaint clusters).   
We explain trends and hotspots of rat complaints, and predict future complaint density or the likelihood of specific areas becoming high-density rat hotspots.   
The predictions would be used ahead of peak seasonal cycles (e.g., warmer months when rat activity spikes) or on a rolling weekly/monthly basis to schedule neighborhood inspections.

**e. Identify potential constraints.**

These could be time limitations, data privacy concerns, regulatory requirements, technical limitations, and ethical considerations.

**Answer:** 
Data Limitations & Biases: data reflects reported complaints rather than actual rat population distribution (e.g., wealthier or more civic-engaged neighborhoods might report complaints at higher rates regardless of actual pest density).

Privacy/Spatial Constraints: Aggregating location data to protect anonymity might obscure exact building-level clusters.

Technical/Time Limitations: Processing large-scale datasets requires to be careful with computing and handling missing or outlier values.
